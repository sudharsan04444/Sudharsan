```python
import pandas as pd
df = pd.read_csv("samplesuperstore.csv")
df.boxplot(column="Sales")
plt.show()
```

```python
df.boxplot(column="Profit")
plt.show()
```

```python
sns.countplot(data =df, x="Category")
plt.title("Number of Orders by Category")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

plt.show()
```

```python
sns.countplot(data =df, x="Segment")
plt.title("Number of Orders by Segment")
plt.xlabel("Segment")
plt.ylabel("Number of Orders")

plt.show()
```

```python
#sales and profit analysis
sns.barplot(data =df, x="Category", y="Sales")
plt.title("Average Sales by Category")
plt.xlabel("Category")
plt.ylabel("Average Sales")

plt.show()
```

```python
#profit by category

sns.barplot(data =df, x="Category", y="Profit")
plt.title("Average Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
#profit distribution

plt.figure(figsize=(8, 6))

sns.boxplot(data=df, x="Category", y="Profit")

plt.title("Distribution of Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
#bivariate analysis
#discount vs profit

plt.figure(figsize=(8, 6))

sns.scatterplot(data=df, x="Discount", y="Profit")

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
#sales vs profit

sns.scatterplot(data=df, x="Sales", y="Profit")
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
#quantity vs sales

plt.figure(figsize=(8, 6))

sns.scatterplot(data=df, x="Quantity", y="Sales")

plt.title("Quantity vs Sales")
plt.xlabel("Quantity")
plt.ylabel("Sales")

plt.show()
```

```python
#multivariate analysis and correlation heatmap

df.info()
```

```python
numeric_cols = [
    "Sales",
    "Quantity",
    "Discount" ,
    "Profit"
]
```

```python
corr = df[numeric_cols].corr()
corr
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    annot=True,
    cmap = "coolwarm"
    )
plt.title("Correlation Heatmap")
plt.show()
```
