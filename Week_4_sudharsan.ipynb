```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df.nlargest(5,"volume")
```

```python
df["daily price change"] = df["close"]- df["open"]
df["daily price change"]
```

```python
df["daily price range"] = df["close"]- df["open"]
df["daily price range"]
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)

plt.title("Trading volume vs daily price change")
plt.xlabel("Date")
plt.ylabel("Trading volume * daily price change")

plt.show()
```

```python
#volume vs daily price range
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price range"
)

plt.title("volume vs daily price range")
plt.xlabel("Date")
plt.ylabel("volume * daily price range")

plt.show()
```

```python
#calculate the 20 day SMA

df["SMA_20"] =df["close"].rolling (window=20).mean()
df["SMA_20"]
```

```python
#calculate the 20 day SMA

df["SMA_50"] =df["close"].rolling (window=50).mean()
df["SMA_50"]
```

```python
from datetime import date
#plot closing price+20_day SMA

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="closing price"
)
sns.lineplot(
    data=df,
    x="date",
    y="SMA_20",
    label="20 day SMA"
)

plt.title("Shopify Closing Price and 20 Day Moving Average")
plt.xlabel("date")
plt.ylabel("price")

plt.show()
```

```python
#plot closing price+_day SMA

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="closing price"
    )

sns.lineplot(
    data=df,
    x="date",
    y="SMA_20",
    label="20 day SMA"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_50",
    label="50 day SMA"
)
plt.title("Shopify Closing Price and 20 and 50 Day Moving Average")
plt.xlabel("date")
plt.ylabel("price")

plt.show()
```

```python
#daily return%

df["daily return%"] = ((df["close"] - df["open"]) / df["open"]) * 100
```

```python
df["daily return%"].head()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
              data=df,
              x="daily return%",
              bins=20,

      )

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return%")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
              data=df,
              x="daily return%",
              bins=20,
              kde=True#kernal density estimate
      )

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return%")
plt.ylabel("Frequency")

plt.show()
```

```python
#volatility

df["Absolute Return"]=df["daily return%"].abs()
df["Absolute Return"]
```

```python
df.nlargest(10,"Absolute Return")
```

```python
plt.figure(figsize=(14,5))

sns.lineplot(
    data=df,
    x="date",
    y="daily return%"
)

plt.axhline(0,linestyle="--")
plt.xlabel("Date")
plt.ylabel("Daily Return%")
plt.title("Shopify daily return over time")

plt.show()
```
